```python
import json
```

```python
link = "data_parser/brands_models.json"

with open(link, "r", encoding="utf-8") as f:
    data = json.load(f)

total_models = sum(len(models) for models in data.values())
total_brands = sum(len(brands) for brands in data.keys())


print(f"All brands: {total_brands}")
print(f"All models: {total_models}")
```
